```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer


df = pd.read_csv('IMDB.csv')


sample_reviews = df['review'].dropna().sample(400).tolist()

# 2. List of target words 
target_words = [
    "masterpiece", "brilliant", "boring", "dull", "good", "amazing",
    "exciting", "satisfying", "bad", "entertaining", "classic", "best",
    "worst", "fantastic", "incredible", "unforgettable", "predictable", "cliché",
     "underrated", "emotional",
]

# Normalize for matching
target_words = [w.lower() for w in target_words]


# 3. Count occurrences of these words

freq = {w: 0 for w in target_words}

for review in sample_reviews:
    text = review.lower()
    for w in target_words:
        if w in text:
            freq[w] += 1


# 4. Plot the results

plt.figure(figsize=(12, 6))
plt.bar(freq.keys(), freq.values(), color='lightcoral')
plt.title("Recurring Words in Movie Reviews")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Bar Chart

The bar chart displays the frequency of specific target words found in a sample of 400 movie reviews from the IMDb dataset, capturing audience sentiments. Words like "masterpiece," "brilliant," and "exciting" reflect positive viewer experiences, while terms like "boring" and "worst" indicate dissatisfaction, providing a clear overview of prevailing opinions on films. Understanding the frequency of these words offers filmmakers and marketers valuable insights into audience preferences and highlights themes that resonate deeply or fail to engage viewers.

```python
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob

df = pd.read_csv('corona_tweets.csv', encoding='latin1')

tweets = df['text'].dropna().sample(100)

scores = tweets.apply(lambda t: TextBlob(t).sentiment.polarity)
labels = scores.apply(lambda s: 'Positive' if s > 0 else 'Negative')

counts = labels.value_counts()

plt.figure(figsize=(7,7))
plt.pie(counts, labels=counts.index, autopct='%1.1f%%')
plt.title("Sentiment of COVID-19 Tweets")
plt.show()
```

Pie Chart

The pie chart illustrates the sentiment distribution of 100 sampled tweets related to COVID-19, categorizing them as either Positive or Negative based on their sentiment polarity. A majority of the tweets are labeled Positive, indicating that many users are expressing supportive or optimistic views around the pandemic, which might reflect community resilience or favorable developments such as vaccine progress. This understanding of public sentiment can help stakeholders, including policymakers and health officials, shape their communication strategies and foster community engagement during the ongoing health crisis.

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import pandas as pd


df = pd.read_csv('tech_news.csv')

df.head()


sample_headlines = df['Title'].dropna().sample(100).tolist()

# Generate word cloud
text = " ".join(sample_headlines)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)

# Display the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Word Cloud from Tech News')
plt.show()
```

Word Cloud

The word cloud visualizes the most frequently mentioned words in a sample of 100 tech news headlines, with larger words indicating higher frequency of occurrence. Terms such as "technology," "innovation," and "AI" prominently stand out, highlighting key themes and topics currently capturing media attention in the tech industry. This visualization effectively summarizes the focal points of recent discourse, providing insights into emerging trends and interests that are shaping the future of technology.
